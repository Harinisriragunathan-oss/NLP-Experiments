```python
import math
import matplotlib.pyplot as plt
sentences = [ "Faculty explains concepts clearly and effectively",
"Hostel food quality needs improvement",
"Placement training sessions are very useful",
"Laboratory computers are outdated",
"Library resources are excellent" ]
words =[]
for sentence in sentences:
  words.extend(sentence.lower().split())
  frequency = {}
  for word in words:
    frequency[word] = frequency.get(word, 0) + 1
    top_words = sorted(frequency.items(),
                       key=lambda x: x[1], reverse=True)[:10]
print("Top Important Words:")
for word, count in top_words:
     print(word, count)
     word_names = [x[0] for x in top_words]
word_scores = [x[1] for x in top_words]
plt.bar(word_names, word_scores)
plt.xlabel("Words")
plt.ylabel("Score")
plt.title("Important Words - TF-IDF")
plt.xticks(rotation=45)
plt.show()

```
