```python
import pandas as pd
```

```python
# Load student feedback dataset
df = pd.read_csv("student_feedback.csv")
```

```python
# Take feedback text
feedback = df["Feedback"].astype(str).tolist()
```

```python
# Convert sentences into words
sentences = [text.lower().split() for text in feedback]
```

```python
from gensim.models import Word2Vec
print("Word2Vec is ready!")
```

```python
# Create Word2Vec model
model = Word2Vec(
sentences, vector_size=50,
window=3,
min_count=1,
workers=4
)
```

```python
# Check similar words
print("Words similar to 'faculty':")
print(model.wv.most_similar("faculty", topn=5))
```

```python
print("\nWords similar to 'placement':")
print(model.wv.most_similar("placement", topn=5))
```
