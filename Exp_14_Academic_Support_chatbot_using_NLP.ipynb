```python
def chatbot(query):
  query = query.lower()
  if "placement" in query:
    return "Placement training and opportunities are available."
  elif "hostel" in query:
      return "Hostel facilities are available for students."
  elif "lab" in query:
    return "Laboratory facilities are available."
  elif "faculty" in query:
    return "Faculty members support students."
  else: return "Sorry, I don't have information about that."
print("Academic Support Chatbot")
print("User: placement")
print("Bot:", chatbot("placement"))
print("User: hostel")
print("Bot:", chatbot("hostel"))
print("User: faculty")
print("Bot:", chatbot("faculty"))
```
