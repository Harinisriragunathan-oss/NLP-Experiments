```python
import pandas as pd
```

```python
df = pd.read_csv("student_feedback.csv")
```

```python
positive = ["excellent", "good", "useful", "clearly", "great"]
negative = ["poor", "bad","outdated", "improvement", "problem"]
```

```python
def sentiment(text):
  text = text.lower()
  pos = sum(word in text for word in positive)
  neg = sum(word in text for word in negative)

  if pos > neg:
    return "Positive"
  elif neg > pos:
    return "Negative"
  else:
    return "Neutral"
df["Sentiment"] = df["Feedback"].apply(sentiment)
print(df[["Feedback", "Sentiment"]])

```
