```python
sentences = [ "placement training is useful", "placement opportunities are increasing" ]
```

```python
# Unigrams
unigrams = []
for sentence in sentences:
  words = sentence.split()
  unigrams.extend(words)
print("Unigrams:")
print(unigrams)
```

```python
# Bigrams

bigrams = []
for sentence in sentences:
  words = sentence.split()
  for i in range(len(words) - 1):
    bigrams.append(words[i] + " " + words[i + 1])
    print("\nBigrams:")
    print(bigrams)
```

```python
 # Bag of Words

 vocabulary = sorted(set(unigrams))
 print("\nVocabulary:")
 print(vocabulary)

 print("\nBag of Words:")
 for sentence in sentences:
  words = sentence.split()
  row = []
 for word in vocabulary:
  row.append(words.count(word))
  print(row)
```
