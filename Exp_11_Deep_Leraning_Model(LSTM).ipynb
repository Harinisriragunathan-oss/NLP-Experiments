```python
import pandas as pd
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
df = pd.read_csv("student_feedback.csv")
texts = df["Feedback"].astype(str)
labels = df["Category"]
classes = {c:i for i, c in enumerate(labels.unique())}
y = np.array([classes[c] for c in labels])
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
X = tokenizer.texts_to_sequences(texts)
X = pad_sequences(X, maxlen=20)
model = Sequential([ Embedding(len(tokenizer.word_index) + 1, 32),
LSTM(32), Dense(len(classes),
activation="softmax") ])
model.compile( optimizer="adam",
loss="sparse_categorical_crossentropy",
metrics=["accuracy"] )
model.fit(X, y, epochs=10, verbose=0)
test = "Hostel food quality is poor."
seq = tokenizer.texts_to_sequences([test])
seq = pad_sequences(seq, maxlen=20)

prediction = model.predict(seq,verbose=0)
category = list(classes.keys())
[np.argmax(prediction)]

print("Input:",test)
print("Category:",category)
```
