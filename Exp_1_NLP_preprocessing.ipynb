```python
 text = "Faculty explains concepts clearly and effectively."
```

```python
#Lowercase
lowercase=text.lower()
print('LOWERCASE')
print(lowercase)

#Tokenization
tokens = word_tokenize(lowercase)
print("\nTOKENIZATION:")
print(tokens)

# Stop-word Removal

print("/nSTOP-WORD REMOVAL:")
stop=["and","the","is","a"]
tokens=[word for word in tokens if
word not in stop]
print(tokens)

#Normalization
print("/nNORMALIZATION")
tokens=[word for word in tokens if
word.isalpha()]
print(tokens)
```
